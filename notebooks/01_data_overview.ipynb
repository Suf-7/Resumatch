# Resumatch: data overview
Quick look at the cleaned data and constructed pairs. Run `make data clean features` first.

```python
import json, sys
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().parent
sys.path.append(str(ROOT / 'data_scripts'))
res = pd.read_parquet(ROOT / 'data/interim/resumes_clean.parquet')
jobs = pd.read_parquet(ROOT / 'data/interim/jobs_clean.parquet')
pairs = pd.read_parquet(ROOT / 'data/processed/pairs.parquet')
res.shape, jobs.shape, pairs.shape
```

```python
for name in ['clean_resumes', 'clean_jobs', 'features']:
    print(name, json.dumps(json.loads((ROOT / f'report/stats/{name}.json').read_text()), indent=1)[:800])
```

```python
from skills import extract_skills
row = pairs[pairs.pair_type == 'hard_negative'].sample(1, random_state=0).iloc[0]
r = res.set_index('resume_id').loc[row.resume_id]
j = jobs.set_index('job_id').loc[row.job_id]
print(r.category, '->', j.title, f'(cos={row.tfidf_cosine})')
print('resume skills:', extract_skills(r.text)[:15])
print('job skills   :', extract_skills(j.text)[:15])
```
